```python
df["Grid Zone"] = [chr(65 + (i % 2)) for i in range(len(df))]
df["Meter ID"] = [f"M{(i %20) + 1:03d}" for i in range(len(df))]
df
```

```python
import itertools

# Household patterns for each zone
patterns = {
    "A": (
        ["H001"] * 2 +
        [f"H{i:03d}" for i in range(2, 9)]
    ),

    "B": (
        ["H010"] * 3 +
        ["H011"] * 2 +
        [f"H{i:03d}" for i in range(12, 17)]
    ),
}

# Assign household IDs within each zone, repeating the pattern
df["Household ID"] = None

for zone, pattern in patterns.items():
    mask = df["Grid Zone"] == zone
    n = mask.sum()
    df.loc[mask, "Household ID"] = list(
        itertools.islice(itertools.cycle(pattern), n)
    )
```

```python
df
```

```python
# rename Electricity Price (USD/kWh) to Tarrif Rate (USD/kWh)
df["Tariff Rate (USD/kWh)"] = df["Electricity Price (USD/kWh)"]
df = df.drop(columns=["Electricity Price (USD/kWh)"])
```

```python
df["Timestamp"] = df["Timestamp"].repeat(20).iloc[:50000].reset_index(drop=True)
df
```

```python
df.to_csv("./smart_grid_dataset_new.csv", index=False)
```
